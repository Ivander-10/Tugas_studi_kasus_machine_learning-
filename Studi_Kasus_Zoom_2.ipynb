```python
import time, math
from multiprocessing import Pool, cpu_count
import numpy as np

def compute_hazard_index(cell_data):
    rain, slope, soil_density = cell_data

    val = 0.0
    for i in range(1, 5000):
        val += math.sin(rain * i) * math.cos(slope / i) + math.sqrt(soil_density)
    return val

if __name__ == '__main__':
    np.random.seed(42)
    n_points = 20000
    grid_data = list(zip(
        np.random.uniform(50, 300, n_points),   # Curah Hujan (mm)
        np.random.uniform(0, 60, n_points),     # Kemiringan Lereng (derajat)
        np.random.uniform(1.1, 2.5, n_points)   # Densitas Tanah (g/cm3)
    ))

    t_start = time.time()
    res_seq = [compute_hazard_index(pt) for pt in grid_data]
    time_seq = time.time() - t_start

    cores = cpu_count()
    t_start = time.time()
    with Pool(processes=cores) as pool:
        res_par = pool.map(compute_hazard_index, grid_data)
    time_par = time.time() - t_start

    print(f"Jumlah CPU Cores Digunakan: {cores}")
    print(f"Waktu Sekuensial : {time_seq:.2f} detik")
    print(f"Waktu Paralel    : {time_par:.2f} detik")
    print(f"Akselerasi Waktu : {time_seq / time_par:.2f}x Lebih Cepat!")
```
